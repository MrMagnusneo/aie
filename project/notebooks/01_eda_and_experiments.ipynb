# Chess Legal Move Classifier: EDA и метрики

Ноутбук показывает базовую проверку self-play данных и метрик. Основной код проекта находится в `src/`.

```python
from pathlib import Path
import json

data_path = Path('../data/generated_self_play.jsonl')
metrics_path = Path('../artifacts/metrics.json')

examples = []
if data_path.exists():
    with data_path.open(encoding='utf-8') as file:
        for line in file:
            examples.append(json.loads(line))

metrics = json.loads(metrics_path.read_text(encoding='utf-8')) if metrics_path.exists() else {}
len(examples), metrics
```

```python
from collections import Counter

label_counts = Counter(example['label'] for example in examples)
position_count = len({example['position_id'] for example in examples})
label_counts, position_count
```

В текущем эксперименте модель сравнивается с tactical baseline по `Top-1 good rate`. Сервис использует только `board.legal_moves`, поэтому нелегальный ход не может быть выбран.
